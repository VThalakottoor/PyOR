```python
T = QS.TransitionState(

    Initial={
        "Manifold": "GS",
        "A1": 0,
        "A2": 0
    },

    Final={
        "Manifold": "ES",
        "B1": 0.5,
        "B2": 0
    }

)
```

```python
T.matrix
```

```python
C = QS.CollapseOperator(

    Initial={
        "Manifold": "ES",
        "B1": 0.5,
        "B2": 0
    },

    Final={
        "Manifold": "GS",
        "A1": 0,
        "A2": 0
    },

    Rate=80e6

)
C.matrix
```
