### Parameters 

```python
# Magnetic Field
B0 = 18.8
offset_N = 0.0
offset_H = gamma("H1","MHz/T") * B0

# Chemical kinetics constant
K_A = 2.0e-10

#pH
pH = -0.8

# Dissociation rates
kd = 10

W = 10**(-pH)/K_A

# J coupling
J_NH = -73.4
J_HH = -16.9

# Relaxation time
T1_N  = 5.0
T1_H1 = 1.0
T1_H2 = 1.0
T1_H3 = 1.0
T1_H4 = 1.0
```

### Define Spin System

```python
QS = QuantumSystems(
    {
        "A": {
            "N":  "N15",
            "H1": "H1",
            "H2": "H1",
            "H3": "H1",
        }
    },
    {
        "B": {
            "N":  "N15",
            "H1": "H1",
            "H2": "H1",
            "H3": "H1",
            "H4": "H1",
        }
    }
)
```

```python
QS.Configure(
    B0=B0,
    PropagationSpace="Liouville",
    MasterEquation="Lindblad",
    Rprocess="Lindblad T1",

    A={
        "OFFSET": {
            "N": offset_N,
            "H1": offset_H,
            "H2": offset_H,
            "H3": offset_H,
        },

        "Jcouplings": [
            ["N",  "H1", J_NH],
            ["N",  "H2", J_NH],
            ["N",  "H3", J_NH],
            ["H1", "H2", J_HH],
            ["H1", "H3", J_HH],
            ["H2", "H3", J_HH],
        ],

        "Lindblad_T1": {
            "N":  5.0,
            "H1": 1.0,
            "H2": 1.0,
            "H3": 1.0,
        },
    },

    B={
        "OFFSET": {
            "N": offset_N,
            "H1": offset_H,
            "H2": offset_H,
            "H3": offset_H,
            "H4": offset_H,
        },

        "Jcouplings": [
            ["N",  "H1", J_NH],
            ["N",  "H2", J_NH],
            ["N",  "H3", J_NH],
            ["N",  "H4", J_NH],
            ["H1", "H2", J_HH],
            ["H1", "H3", J_HH],
            ["H1", "H4", J_HH],
            ["H2", "H3", J_HH],
            ["H2", "H4", J_HH],
            ["H3", "H4", J_HH],
        ],

        "Lindblad_T1": {
            "N":  5.0,
            "H1": 1.0,
            "H2": 1.0,
            "H3": 1.0,
            "H4": 1.0,
        },
    },
)
```

### Hamiltonian

```python
# Zeeman for systemv A
Hz_A = -QS.A.OFFSET["N"] * QS.A.Nz - QS.A.OFFSET["H1"] * (QS.A.H1z + QS.A.H2z + QS.A.H3z)
#Hz_A.matrix
```

```python

# Zeeman for systemv B
Hz_B = -QS.B.OFFSET["N"] * QS.B.Nz - QS.B.OFFSET["H1"] * (QS.B.H1z + QS.B.H2z + QS.B.H3z  + QS.B.H4z)
#Hz_B.matrix
```

```python
HJ0_NH3 = J_HH * (
      QS.A.H1x * QS.A.H2x
    + QS.A.H1y * QS.A.H2y
    + QS.A.H1z * QS.A.H2z

    + QS.A.H1x * QS.A.H3x
    + QS.A.H1y * QS.A.H3y
    + QS.A.H1z * QS.A.H3z

    + QS.A.H2x * QS.A.H3x
    + QS.A.H2y * QS.A.H3y
    + QS.A.H2z * QS.A.H3z
)

HJ1_NH3 = J_NH * (
      QS.A.Nz * QS.A.H1z
    + QS.A.Nz * QS.A.H2z
    + QS.A.Nz * QS.A.H3z
)

Hj_A = HJ0_NH3 + HJ1_NH3
#Hj_A.matrix
```

```python
HJ0_NH4 = J_HH * (
      QS.B.H1x * QS.B.H2x
    + QS.B.H1y * QS.B.H2y
    + QS.B.H1z * QS.B.H2z

    + QS.B.H1x * QS.B.H3x
    + QS.B.H1y * QS.B.H3y
    + QS.B.H1z * QS.B.H3z

    + QS.B.H1x * QS.B.H4x
    + QS.B.H1y * QS.B.H4y
    + QS.B.H1z * QS.B.H4z

    + QS.B.H2x * QS.B.H3x
    + QS.B.H2y * QS.B.H3y
    + QS.B.H2z * QS.B.H3z

    + QS.B.H2x * QS.B.H4x
    + QS.B.H2y * QS.B.H4y
    + QS.B.H2z * QS.B.H4z

    + QS.B.H3x * QS.B.H4x
    + QS.B.H3y * QS.B.H4y
    + QS.B.H3z * QS.B.H4z
)


HJ1_NH4 = J_NH * (
      QS.B.Nz * QS.B.H1z
    + QS.B.Nz * QS.B.H2z
    + QS.B.Nz * QS.B.H3z
    + QS.B.Nz * QS.B.H4z
)


Hj_B = HJ0_NH4 + HJ1_NH4

#Hj_B.matrix
```

```python
Htotal_A = 2.0 * QS.A.pi * (Hz_A + Hj_A) 
Htotal_B = 2.0 * QS.A.pi * (Hj_B) 

Htotal_A_L = QS.A.QuantumLibrary.CommutationSuperoperator(Htotal_A)
Htotal_B_L = QS.B.QuantumLibrary.CommutationSuperoperator(Htotal_B)
```

### Partial Trace Superoperator

```python
# Partial-trace superoperators: NH4 -> NH3

Tr1 = QS.B.Evolutions.PartialTrace_Superoperator(
    keep=[0, 2, 3, 4],
    QuantumObject=True
)

Tr2 = QS.B.Evolutions.PartialTrace_Superoperator(
    keep=[0, 1, 3, 4],
    QuantumObject=True
)

Tr3 = QS.B.Evolutions.PartialTrace_Superoperator(
    keep=[0, 1, 2, 4],
    QuantumObject=True
)

Tr4 = QS.B.Evolutions.PartialTrace_Superoperator(
    keep=[0, 1, 2, 3],
    QuantumObject=True
)

Tr = (Tr1 + Tr2 + Tr3 + Tr4) / 4.0
```

### Kronecker Superoperator

```python
rho_H = QS.A.Eye(2) / 2.0

Kron1 = QS.A.Evolutions.Kronecker_Superoperator(
    rho_H,
    insert_at=1
)

Kron2 = QS.A.Evolutions.Kronecker_Superoperator(
    rho_H,
    insert_at=2
)

Kron3 = QS.A.Evolutions.Kronecker_Superoperator(
    rho_H,
    insert_at=3
)

Kron4 = QS.A.Evolutions.Kronecker_Superoperator(
    rho_H,
    insert_at=4
)

Kron = (Kron1 + Kron2 + Kron3 + Kron4) / 4.0
```
